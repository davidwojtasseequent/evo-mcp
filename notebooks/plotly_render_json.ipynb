## Test plotly generation tool functionality from Evo MCP

The Evo-MCP server has a number of tools to generate plotly-compatible JSON data for common plots. This notebook demonstrates how to use these tools to 
- create a grade histogram from downhole assay data.

```python
import sys
from pathlib import Path
import plotly.io as pio
import json

# Add src directory to Python path
repo_root = Path.cwd().parent
src_path = repo_root / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from evo_mcp.utils import generate_grade_histogram, get_downhole_collection, download_interval_data, generate_grade_violin
```

```python
workspace_id = "5c20f625-6626-4a09-a9ea-d6534e8d0407" # demo1
object_id = "3ac1f7d5-687f-4515-813e-edec3cb4f4e2"  # downhole collection OLstage1
version = ""

obj, obj_dict = await get_downhole_collection(workspace_id, object_id, version)
```

```python
collection_name = "assay"
df = await download_interval_data(obj, collection_name)
```

```python
column_name = "Au"
plotly_data = generate_grade_histogram(df, column_name, bins=30)

fig = pio.from_json(json.dumps(plotly_data))
fig.show()
```

## Violin Plot Example

Violin plots show the full distribution of grade values with quartiles and mean lines. They're excellent for comparing distributions across different holes or zones.

```python
# Single violin for entire dataset
violin_data = generate_grade_violin(df, column_name)

fig_violin = pio.from_json(json.dumps(violin_data))
fig_violin.show()
```

```python
# Violin plot grouped by hole_id (shows distribution for each hole)
violin_grouped = generate_grade_violin(df, column_name, group_by="hole_id")

fig_violin_grouped = pio.from_json(json.dumps(violin_grouped))
fig_violin_grouped.show()
```
